```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate 1,000 values from a normal distribution
rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=10, size=1000)

# Create a histogram
plt.hist(data, bins=30, color="steelblue", edgecolor="white")
plt.title("Histogram of Simulated Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
